# Processing - Climate Indices
- 해양 지역의 기상학적 변수를 기반으로 계산되는 전지구 단위 기후 인덱스
- 따라서 시계열 자료 형태임.
## Variables
- NINO 3.4 : 태평양 특정 해역(5N-5S, 170W-120W)의 SST Anomaly
- IOD (Indian Ocean Dipole) : West Pole SST Anomaly (50E-70E, 10S-10N) 에서 East Pole SST Anomaly (90E-110E, 10S-0S) 를 뺀 값
- WVG (Western V Gradient) : 표준화된 WP SST Anomaly에서 표준화된 Nino 4 SST Anomaly를 뺀 값
  - Western Pacific (WP) : 130E-150E, 0-20N
  - Nino 4 (Central Pacific) : 160E-150W, 5S-5N
  - WVG는 '표준화(Z-score)'된 값의 차이를 사용함(Funk et al.)
  - WVG가 높음(+) -> 서태평양 온난, 중태평양 한랭(La Nina 경향) -> 동아프리카 가뭄 유발
- MEI (Multivariate ENSO Index) : 5가지 핵심 기상 변수(SST, SLP, U, V, OLR)를 복합적으로 결합하여 산출
  - Positive (+) : 값이 클수록 강력한 엘니뇨를 의미함. 전 지구적 기온 상승과 연관되며, IOD와 상호작용하여 동아프리카에 이상 기후를 유발함.
  - Negative (-) : 값이 작을수록 강력한 라니뇨를 의미함. 동아프리카 지역에 극심한 가뭄을 일으키는 주범으로 작용하는 경우가 많음.

```python
import os
from google.colab import drive

drive.mount('/content/drive')

folder_path = '/content/drive/MyDrive/WFP Collab/subnational_price_prediction/data/climate_indices/raw'
file_nm1 = 'Climate_Indices_OISST_2007_2025.csv'
file_nm2 = 'MEI_v2_Indices_2007_2025.csv'

full_path1 = os.path.join(folder_path, file_nm1)
full_path2 = os.path.join(folder_path, file_nm2)
```

```python
from pathlib import Path

output_folder = Path('/content/drive/MyDrive/WFP Collab/subnational_price_prediction/data/climate_indices/processed')
output_folder.mkdir(parents=True, exist_ok=True)

output_path = output_folder / 'Climate_Indices_2007_2025.csv'
print(output_path.exists())
```

```python
import pandas as pd
```

```python
df_oisst = pd.read_csv(full_path1)
df_oisst.head()
```

```python
df_oisst['year'] = df_oisst['date'].str[:4].astype(int)
df_oisst_f = df_oisst[['year', 'month', 'NINO34_Anom', 'IOD_DMI', 'Western_V_Gradient']]
df_oisst_f.head()
```

```python
pd.DataFrame(df_oisst_f.value_counts('month')).T
```

```python
df_mei = pd.read_csv(full_path2)
df_mei.head()
```

```python
pd.DataFrame(df_mei.value_counts('Year')).T
```

```python
pd.DataFrame(df_mei.value_counts('Month')).T
```

```python
df_merged = pd.merge(df_oisst_f, df_mei, left_on=['year', 'month'], right_on=['Year', 'Month'], how='left').drop(columns=['date', 'Year', 'Month'])
df_merged.head()
```
