```python
import pandas as pd

# 파일 경로
file_path = '../../data/population/eth_admpop_2023.xlsx'

# Dataframe 로드 (Admin2 시트)
df = pd.read_excel(file_path, sheet_name='ETH_admpop_adm2_2023')

# 필요한 컬럼만 선택 (Metadata + Population)
target_cols = [
    'admin2Name_en', 'admin2Pcode', 
    'admin1Name_en', 'admin1Pcode', 
    'admin0Name_en', 'admin0Pcode', 
    'Total'
]
df = df[target_cols]

# Admin2 별로 Total 인구 합계 (이미 Admin2 단위 시트면 sum 불필요하지만 안전하게)
df_pop = df.groupby(target_cols[:-1])['Total'].sum().reset_index()

# 컬럼명 변경 (Total -> population)
df_pop.rename(columns={'Total': 'population'}, inplace=True)

df_pop.head()
```

```python
import pandas as pd
import numpy as np

# 파일 경로
file_path_som = '../../data/population/somalia_population.xlsx'

# 데이터 로드 (헤더는 1번째 줄이지만 데이터와 매핑이 어긋나므로 header=None으로 읽고 처리 권장, 
# 하지만 편의상 default로 읽고 컬럼 위치로 접근합니다)
df_som = pd.read_excel(file_path_som)
```

```python

# 컬럼 위치 기반 추출
# Col 2 (idx 2): Mixed Region Name & District P-Code
# Col 3 (idx 3): District (Admin2) Name
# Last Col (idx -1): Total Population
df_som_final = df_som[['Region','District','P_Code', 'Total']]
df_som_final.head()
```

# Crop calendar

```python
df_som['raw_region'] = df.iloc[:, 2]
df_som['admin2'] = df.iloc[:, 3]
df_som['population'] = df.iloc[:, -1]

# Admin1 (Region) 추출 로직: 'SO'로 시작하지 않는 값이 Region Name -> Forward Fill
is_pcode = df_som['raw_region'].astype(str).str.startswith('SO')
df_som['admin1'] = df_som['raw_region'].where(~is_pcode).ffill()

# Admin2 Pcode 추출: 'SO'로 시작하는 값만 남김 (Region Name으로 된 행은 Pcode 결측 처리됨)
df_som['admin2Pcode'] = df_som['raw_region'].where(is_pcode)

# 메타데이터 추가
df_som['admin0Name_en'] = 'Somalia'
df_som['admin0Pcode'] = 'SO'

# 불필요한 행 제거 (헤더 행 등)
df_som = df_som.dropna(subset=['admin2'])
df_som = df_som[df_som['admin2'] != 'District'] # 헤더 잔여물 제거

# 숫자형 변환
df_som['population'] = pd.to_numeric(df_som['population'], errors='coerce')

# 최종 컬럼 선택
cols = ['admin2', 'admin2Pcode', 'admin1', 'admin0Name_en', 'admin0Pcode', 'population']
df_som_final = df_som[cols].dropna(subset=['population'])

df_som_final.head()
```
